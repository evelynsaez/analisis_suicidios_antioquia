<center><img src='https://es-static.z-dn.net/files/d1f/9d919bd4ba4c9f82673b6935e2dcf657.png' width = '50%'></center>
Solicite estas distancias e indique la distancia calculada.


````python
import math
import numpy as nup
#Captura de Datos
desp_horizontal = float(input('Ingrese el valor del desplazamiento horizontal: '))
desp_vertical = float(input('Ingrese el valor del desplazamiento vertical: '))

#Cálculos
discriminante = math.pow(desp_horizontal, 2) + math.pow(desp_vertical, 2)
dist_recta = math.sqrt(discriminante)

#Visualización
print(f'''
Para un desplazamiento:
Horizontal      = {desp_horizontal:.2f}
Vertical        = {desp_vertical:.2f}

Se obtiene:
Distancia Recta = {dist_recta:.2f}
''')
````

````python
#Es mejor hacer el mapa primero de Poligonos, lineas y puntos

m = comunas.explore( #Trae algunos datos y muestre que tengo disponible
    column='NOMBRE', #La columna que quiero mirar es la de nombre
    tooltip='NOMBRE', #Sirve para crear un mensaje al pasar el puntero por el nombre de la comuna
    #Se puede hacer un nuevo dataframe para que muestre solo los datos que queramos
    popup=True, #Activa la ventana emergente para cuando clickeo aparece los atributos
    cmap="tab20c", #Se puede cambiar las paletas de color del mapa
    legend=False, #Muestra leyenda con valores unicos (Se pude activar true o desactivar false)
    style_kwds=dict(color="black", weight=1), #Mostrar el estilo y contorno del mapa
    name="Comunas" #Se muestra una capa de comunas para que se active desactive
)

m = estaciones.explore(
    column='NOMBRE',
    tooltip='NOMBRE',
    popup=True,
    cmap='tab20c',
    legend=False,
    marker_kwds=dict(radius=8, fill=True), #Define el radio de los puntos y el lleno es verdad para que llene el circulo
    style_kwds=dict(color="black", weight=1),
    name='Estaciones de Metro',
    m = m, #Para evitar que el mapa se sobreescriba le pongo m = m
)

# Asegura tener un mapa de base
folium.TileLayer("Stamen Terrain", #nombre del mapa base que se va a utilizar
                 show=True, #le decimos que queremos mostrar el mapa base
                 #Mostrar la licencia y derechos de autor y es obligatorio
                 attr='Map tiles by <a href="http://stamen.com">Stamen Design</a>, <a href="http://creativecommons.org/licenses/by/3.0">CC BY 3.0</a> &mdash; Map data &copy; <a href="https://www.openstreetmap.org/copyright">OpenStreetMap</a> contributors'
).add_to(m) #se le debe agregar al mapa con este comando

# use folium to add layer control
folium.LayerControl().add_to(m) #Prende y apaga el control de capas (comunas y estaciones) y se agrega al mapa

# Guardar archivo html
m.save('mapa.html')

# Mostrar el mapa
m
````
